Config Loader

```python
import json

CONFIG_PATH = "/lakehouse/default/Files/config/pipeline_config.json"

def load_config():
    with open(CONFIG_PATH, "r") as f:
        return json.load(f)

def get_source(config, source_name):
    return config["sources"][source_name]

def get_last_loaded(config, key):
    return config["last_loaded"][key]

def update_last_loaded(config, key, value):
    config["last_loaded"][key] = value
    with open(CONFIG_PATH, "w") as f:
        json.dump(config, f, indent=2)
```

```python
config = load_config()
print(config["lakehouse"]["name"])
print(get_last_loaded(config, "atfm_delay_year"))
```
