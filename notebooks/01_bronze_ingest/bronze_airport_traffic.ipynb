```python
import uuid
from datetime import datetime

config = load_config()
source = get_source(config, "airport_traffic")
last_year = get_last_loaded(config, "airport_traffic_year")

current_year = 2024
batch_id = str(uuid.uuid4())

for year in range(last_year + 1, current_year + 1):
    url = source["url_pattern"].replace("{year}", str(year))

    count = ingest("airport_traffic", url, "csv", "airport_traffic", batch_id)

    update_last_loaded(config, "airport_traffic_year", year)

    print(f"Year {year}: {count} rows loaded")
```
