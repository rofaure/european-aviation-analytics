```python
import uuid
from datetime import datetime

config = load_config()
source = get_source(config, "opdi_flight_list")
last_month = get_last_loaded(config, "opdi_month")

start_year = int(last_month[:4])
current_year = 2024
batch_id = str(uuid.uuid4())

for year in range(start_year + 1, current_year + 1):
    for month in range(1, 13):
        yyyymm = f"{year}{month:02d}"
        url = source["url_pattern"].replace("{YYYYMM}", yyyymm)

        count = ingest("opdi_flight_list", url, "parquet", "opdi_flight_list", batch_id)

        update_last_loaded(config, "opdi_month", yyyymm)
        print(f"{yyyymm}: {count} rows loaded")
```
