```python
from pyspark.sql import functions as F

dim_airlines_df = spark.read.format("delta").load(
    "abfss://EAA_WorkSpace@onelake.dfs.fabric.microsoft.com/EAA_LakeHouse.Lakehouse/Tables/silver/airline"
)

columns = [
    "icao_code",
    "iata_code",
    "airline_name",
    "carrier_type",
    "country",
]
```

```python
# Check 1: count nulls in every column.
null_counts = (
    dim_airlines_df.agg(
        *[F.sum(F.col(c).isNull().cast("int")).alias(c) for c in columns]
    )
    .first()
    .asDict()
)

# Check 2: find duplicated non-null values in each code column.
duplicate_icao = (
    dim_airlines_df.filter(F.col("icao_code").isNotNull())
    .groupBy("icao_code")
    .count()
    .filter(F.col("count") > 1)
)

duplicate_iata = (
    dim_airlines_df.filter(F.col("iata_code").isNotNull())
    .groupBy("iata_code")
    .count()
    .filter(F.col("count") > 1)
)

duplicate_counts = {
    "icao_code": duplicate_icao.count(),
    "iata_code": duplicate_iata.count(),
}

print(f"Null counts: {null_counts}")
print(f"Duplicate code counts: {duplicate_counts}")

if duplicate_counts["icao_code"]:
    display(duplicate_icao.orderBy(F.desc("count")))

if duplicate_counts["iata_code"]:
    display(duplicate_iata.orderBy(F.desc("count")))
```
